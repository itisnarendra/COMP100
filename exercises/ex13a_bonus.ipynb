# Zoo Modelling Exercise

## Overview
In this assignment, you will model a **Zoo system** using:
-   Classes
-   Inheritance
-   Composition
-   File I/O

### System Structure
-   A `Zoo` contains multiple (dict) of `Pavilions`
-   A `Pavilion` will house multiple (list) of `Animals`
-   Animals are categorized as:
    -   `Bird`
    -   `Reptile`
    -   `Mammal`

We will follow a **bottom-up design approach**:
> Start by implementing the smallest building blocks (Bird, Reptile and Mammal) and then build upwards.   
> However, in this case it is advisable to start with the Animal class

[The other option is top-down design]
> Start by implementing the largest building blocks (Zoo) and then build downwards.   

## 1.0 Animal (Base Class)

### Purpose
This class represents a general animal.

> Students should not normally create Animal objects directly in a real-world system. Hence an abstract would be preferred here. 
> Animal serves as the parent class for Bird, Reptile, and Mammal. 

<!-- Because Python does not have a straight forward way of implementing Abstract Class -->

### Attributes
-   `name`   
    a string representing the common name of this animal *(e.g. Polar Bear, Barn Owl)*
-   `native`   
    a string representing the origin of this animal *(e.g. Canada, Australia)*

### Methods
-   `__init__(self, name, native)`   
    -   This is the constructor.  
    -   It assigns the argument to instance variables of matching names

-   `__str__(self)`   
    Returns a string that includes all the instance variables of this objects.  
    See sample output to write this method.


### Test Harness
```python
a = Animal('Toucan', 'Guyana')      #constructor
print(a)                            #__str__ method
b = Animal('Tortoise', 'Galapagos')
print(b)
c = Animal('Kangaroo', 'Australia')
print(c)
```

#### Sample output
```text
I am a(n) Toucan and I come from Guyana.
I am a(n) Tortoise and I come from Galapagos.
I am a(n) Kangaroo and I come from Australia.
```

## 1.1 Bird (Subclass of Animal)
### Purpose
This is a child class of `Animal`.

### Attributes
In addition of the `name` and `native` attribute of the parent class, the `Bird` class will also have the following attribute
-   `can_fly`   
    a bool indicating if this animal is capable of flight

### Methods
-   `__init__(self, name, native, can_fly = True)`   
    -   This is the constructor.  
    -   It invokes the parent constructor (using `super()`) with the first two arguments. 
    -   It assigns the third argument to an instance variable

-   `__str__(self)`   
    Returns a string that includes all the instance variable of this objects.  
    See sample output to write this method.


### Test Harness
```python
d = Bird('Peregrine Falcon', 'Peru')                        #constructor with default can_fly value
print(d)                                                    #__str__ method
e = Bird('Emperor Penguin', 'Antarctica', can_fly=False)    #constructor with can_fly set to False
print(e)
f = Bird('Ostrich', 'Africa', can_fly=False)
print(f)
```

#### Sample output

## 2.0 Pavilion 
### Purpose
This class will compose of objects of the `Animal` class.

### Attributes
There are two attributes in this class:
-   `name`   
    a string representing the name of this pavilion.
-   `animals`   
    a list (of animals) indicating the animal objects housed in this pavilion.

### Methods
-   `__init__(self, name)`   
    -   This is the constructor.  
    -   It assigns the argument to an instance variable
    -   It creates an instance variable to store the intended list of animals. 

-   `add_animal(self, animal)`  
    -   This method takes a single argument representing an animal object.
    -   It appends it to the list variable `animals`.
    -   It does not explicilty returns anything.
    -   It does not display anything. 

-   `get_animals(self)`   
    -   This method does not take an explicit argument.
    -   It returns a single string representing the names of all the animals in this pavilion separated by a comma and a space.
    -   It is use in the `__str__` method described below.
    -   It does not display anything. 

-   `__str__(self)`   
    Returns the name of this pavilion and the names of all the animals contained.  
    See sample output to write this method.


### Test Harness
```python
australian = Pavilion('Australian')             #constructor
australian.add_animal(c)                        #add_animal method
australian.add_animal(e)
australian.add_animal(h)
australian.add_animal(k)
print(australian)                               #__str__ method and get_animals method

south_america = Pavilion('South American')
south_america.add_animal(a)
south_america.add_animal(d)
south_america.add_animal(g)
south_america.add_animal(j)
print(south_america)

african = Pavilion('African')
african.add_animal(f)
african.add_animal(i)
african.add_animal(m)
african.add_animal(n)
print(african)
```

#### Sample output
```text
Welcome to the Australian Pavilion! We have the following animals: Kangaroo, Emperor Penguin, Alligator, Koala.
Welcome to the South American Pavilion! We have the following animals: Toucan, Peregrine Falcon, Caiman, Capybara.
Welcome to the African Pavilion! We have the following animals: Ostrich, Boa constrictor, Giraffe, Lowland Gorilla.
```

## 3.0 Zoo 
### Purpose
This class will compose of objects of the `Pavilion` class.

### Attributes
There are two attributes in this class:
-   `name`   
    a string representing the name of this zoo.
-   `pavilions`   
    a dict of pavilion name and pavilion housed in this zoo.

### Methods
-   `__init__(self, name)`   
    -   This is the constructor.  
    -   It assigns the argument to an instance variable
    -   It creates an instance variable to store the intended dict of pavilions. 

-   `add_pavilion(self, pavilion)`  
    -   This method takes a single arguments representing a pavilion object and does the following:
        -   Adds a key-value pair to the instance variable pavilions.
            -   The key is the name of the pavilion and the value is the pavilion object.
    -   It does not return anything. 
    -   It does not display anything. 

-   `add_animal_to_pavilion(self, animal, pavilion_name)`  
    -   This method takes two arguments an animal object and a string representing the pavilion name and does the following:
        -   Uses the dict `pavilion` and the key `pavilion_name` to obtain the pavilion object.
            -   It then uses the add_animal of the pavilion object to add the animal object to it.
    -   It does not return anything. 
    -   It does not display anything. 

-   `print(self)`   
    -   This method does not take an explicit argument.
    -   It output a representation of this zoo object.
    -   See the sample output.
    -   It does not return anything. 

-   `save(self, filename)`   
    Saves this zoo information to the specified text.  
    See sample output to write this method.


### Test Harness
```python
zoo = Zoo('Toronto')                            #zoo constructor
zoo.add_pavilion(australian)                    #add_pavilion method of zoo class
zoo.add_pavilion(south_america)
zoo.add_pavilion(african)

zoo.print()                                     #print(zoo)

lion = Mammal('Lion', 'Africa')                 #mammal constructor
zoo.add_animal_to_pavilion(lion, 'African')     #add_animal_to_pavilion method of zoo class
zoo.print()

zoo.save('zoo_info.txt')                        #save method of zoo class
```
